# DATA STRUCTURES - LIST, DICTIONARY, SET & CONDITIONAL STATEMENTS

**List Creation**

```python
# Create age_list with five integer elements

age_list = [24, 25, 26, 27, 28]
print("Age List:", age_list)
```

```python
# Create name_list with five string elements

name_list = ["Sathish", "Pallavi", "Leo", "Hemanth", "Prince"]
print("Name List:", name_list)
```

**List Operations / Modifications**

```python
# 2(a). Append Yazhini to name_list

name_list.append("Yazhini")
print("After Append:", name_list)
```

```python
# 2(b). Insert 30 at index 2 in age_list

age_list.insert(2, 30)
print("After Insert:", age_list)
```

```python
# 2(c). Remove Yazhini from name_list

name_list.remove("Yazhini")
print("After Remove:", name_list)
```

```python
# 2(d). Pop the last element from age_list

age_list.pop()
print("After Pop:", age_list)
```

```python
# 2(e). Extend age_list with additional ages

age_list.extend([29, 30, 26])
print("After Extend:", age_list)
```

```python
# 2(f). Sort age_list in descending order

age_list.sort(reverse=True)
print("Descending Order:", age_list)
```

```python
# 2(g). Find maximum, minimum and sum
print("Maximum Age:", max(age_list))
print("Minimum Age:", min(age_list))
print("Sum of Ages:", sum(age_list))
```

**Accessing List Elements**

```python
# 3(a). Print first element

print("First Element:", name_list[0])
```

```python
# 3(b). Print last element

print("Last Element:", name_list[-1])
```

```python
# 3(c). Print elements from index 2 to index 4

print("Elements from Index 2 to 4:", name_list[2:5])
```

```python
# 3(d). Print name_list in reverse order

print("Reverse List:", name_list[::-1])
```

**Dictionary Creation, Modification and Access**

```python
# 4(a). Create student_marks dictionary
student_marks = {
    "Sathish": 85,
    "Pallavi": 90,
    "Leo": 75,
    "Hemanth": 88,
    "Prince": 95
}

print("Student Marks:", student_marks)
```

```python
# Access and print a specific student's mark

print("Sathish Mark:", student_marks["Sathish"])
```

```python
# 4(c). Add a new student Janani with mark 80

student_marks["Janani"] = 80
print("After Adding Janani:", student_marks)
```

```python
# Update an older student's mark to 82

student_marks["Hemanth"] = 82
print("After Updating Hemanth:", student_marks)
```

**Sets Operations**

```python
# 5(a). Create a set

my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print("My Set:", my_set)
```

Explanation: A set automatically removes duplicate values. Therefore, the repeated 'a' and 'i' appear only once. Sets are unordered, so the output order may vary.

```python
# 5(b). Attempt to change a set element

try:
    my_set[4] = 's'
except TypeError as e:
    print("Error:", e)
```

Explanation: Sets do not support indexing or item assignment because they are unordered collections. You can add or remove elements using add() and remove().

```python
# 5(c). Create two sets

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
```

```python
# 5(d). Find union and intersection

print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

**Operators & Conditional Statements (IF, ELIF, ELSE)**

```python
# Performance Category Program

score = int(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent performance! Keep it up.")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
